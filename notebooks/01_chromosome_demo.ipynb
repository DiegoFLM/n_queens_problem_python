# Chromosome Class Demonstration

This notebook demonstrates the `Chromosome` class, which represents a candidate solution to the N-Queens problem.

## Overview

The Chromosome class uses **permutation encoding**:
- Index represents the column position
- Value represents the row position of the queen
- This encoding guarantees no two queens share the same row or column
- Only diagonal conflicts need to be checked

```python
# Add parent directory to path to import our package
import sys
sys.path.insert(0, '..')

import numpy as np
from src.n_queens import Chromosome, visualization
```

## Creating a Chromosome

A chromosome can be created from any array-like object. Let's create a simple 4-Queens configuration:

```python
# Create a 4-Queens chromosome
positions = np.array([1, 3, 0, 2])
chromosome = Chromosome(positions)

print(f"Board size: {chromosome.get_n()}")
print(f"Queen positions: {chromosome.get_positions()}")
```

## Visualizing the Board

We can visualize the chessboard configuration:

```python
# Text representation
chromosome.print_board()
```

```python
# Visual representation with queen symbols
visualization.render_board(chromosome.get_positions(), title="4-Queens Example")
```

## Generating Random Chromosomes

The `random_vec()` method creates random valid configurations:

```python
# Create a base chromosome
base = Chromosome(np.arange(8))

# Generate a random 8-Queens configuration
random_positions = base.random_vec()
random_chromosome = Chromosome(random_positions)

print(f"Random configuration: {random_chromosome.get_positions()}")
visualization.render_board(random_chromosome.get_positions(), title="Random 8-Queens Configuration")
```

## Mutation: Creating Offspring

The `make_child()` method implements swap mutation - it creates a new chromosome by swapping two queen positions:

```python
# Create parent chromosome
parent = Chromosome(np.array([3, 1, 6, 2, 5, 7, 4, 0]))
print(f"Parent:  {parent.get_positions()}")
print(f"Parent fitness: {parent.fitness():.3f}")

# Create child through mutation
child = parent.make_child()
print(f"\nChild:   {child.get_positions()}")
print(f"Child fitness: {child.fitness():.3f}")
```

## Fitness Evaluation

The fitness function measures solution quality:
- **Fitness = 1.0**: Perfect solution (no conflicts)
- **Fitness < 1.0**: Has diagonal conflicts
- Based on number of mutual threats (diagonal attacks)

```python
# Example 1: Poor solution
poor_solution = Chromosome(np.array([0, 1, 2, 3, 4, 5, 6, 7]))
print("Poor solution (diagonal line):")
print(f"  Positions: {poor_solution.get_positions()}")
print(f"  Mutual threats: {poor_solution.mutual_threats()}")
print(f"  Fitness: {poor_solution.fitness():.3f}")

# Example 2: Perfect solution
perfect_solution = Chromosome(np.array([3, 6, 2, 7, 1, 4, 0, 5]))
print("\nPerfect solution:")
print(f"  Positions: {perfect_solution.get_positions()}")
print(f"  Mutual threats: {perfect_solution.mutual_threats()}")
print(f"  Fitness: {perfect_solution.fitness():.3f}")
```

## Comparing Multiple Configurations

Let's compare several random configurations for an 8-Queens problem:

```python
# Generate and evaluate multiple random configurations
base = Chromosome(np.arange(8))
configurations = []

print("Randomly generated 8-Queens configurations:\n")
for i in range(5):
    chrom = Chromosome(base.random_vec())
    configurations.append(chrom)
    print(f"Config {i+1}: {chrom.get_positions()} - "
          f"Threats: {chrom.mutual_threats()}, "
          f"Fitness: {chrom.fitness():.3f}")
```

## Finding Solutions Through Random Search

Let's try to find a valid solution through random generation (simple Las Vegas approach):

```python
# Search for a solution
base = Chromosome(np.arange(8))
attempts = 0
max_attempts = 10000

print("Searching for a valid 8-Queens solution...\n")

while attempts < max_attempts:
    candidate = Chromosome(base.random_vec())
    attempts += 1
    
    if candidate.fitness() == 1.0:
        print(f"✓ Solution found after {attempts} attempts!")
        print(f"  Positions: {candidate.get_positions()}")
        visualization.render_board(
            candidate.get_positions(),
            title=f"8-Queens Solution (found in {attempts} attempts)"
        )
        break
else:
    print(f"No solution found in {max_attempts} attempts")
```

## Summary

The Chromosome class provides:
- **Permutation encoding** for valid N-Queens configurations
- **Random generation** for creating diverse populations
- **Mutation operator** for evolutionary algorithms
- **Fitness evaluation** to measure solution quality

These building blocks enable various algorithmic approaches including genetic algorithms, Las Vegas, and Monte Carlo methods.
