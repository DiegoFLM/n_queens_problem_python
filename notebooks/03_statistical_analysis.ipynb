# Statistical Analysis of N-Queens Genetic Algorithm

This notebook provides statistical analysis of the genetic algorithm's performance, including:
- Population fitness distribution
- Convergence analysis
- Performance metrics across different configurations

## Setup

```python
# Add parent directory to path
import sys
sys.path.insert(0, '..')

import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
from src.n_queens import Chromosome, EvolManager, visualization

# Set style
sns.set_style("whitegrid")
plt.rcParams['figure.figsize'] = (12, 6)

np.random.seed(42)
```

## Initial Population Analysis

Let's analyze the fitness distribution of a randomly initialized population.

```python
# Create a large population for 12-Queens
manager = EvolManager(
    genes_per_chrom=12,
    pop=1000,
    generations=200,
    reproductive_pool_size=800,
    offspring=500
)

# Get initial population
initial_population = manager.get_chrom_array()

# Extract fitness values
initial_fitness = [chrom.fitness() for chrom in initial_population]

print(f"Initial Population Statistics (N=12, Population=1000)")
print(f"{'='*60}")
print(f"Mean fitness:   {np.mean(initial_fitness):.4f}")
print(f"Std deviation:  {np.std(initial_fitness):.4f}")
print(f"Min fitness:    {np.min(initial_fitness):.4f}")
print(f"Max fitness:    {np.max(initial_fitness):.4f}")
print(f"Median fitness: {np.median(initial_fitness):.4f}")
```

```python
# Visualize initial fitness distribution
fig, (ax1, ax2) = plt.subplots(1, 2, figsize=(14, 5))

# Histogram
ax1.hist(initial_fitness, bins=30, color='#2E86AB', alpha=0.7, edgecolor='black')
ax1.axvline(np.mean(initial_fitness), color='red', linestyle='--', linewidth=2, label='Mean')
ax1.axvline(np.median(initial_fitness), color='green', linestyle='--', linewidth=2, label='Median')
ax1.set_xlabel('Fitness', fontsize=12, weight='bold')
ax1.set_ylabel('Frequency', fontsize=12, weight='bold')
ax1.set_title('Initial Population Fitness Distribution', fontsize=14, weight='bold')
ax1.legend()
ax1.grid(True, alpha=0.3)

# KDE plot
sns.kdeplot(data=initial_fitness, ax=ax2, fill=True, color='#2E86AB', alpha=0.5)
ax2.set_xlabel('Fitness', fontsize=12, weight='bold')
ax2.set_ylabel('Density', fontsize=12, weight='bold')
ax2.set_title('Fitness Density Distribution', fontsize=14, weight='bold')
ax2.grid(True, alpha=0.3)

plt.tight_layout()
plt.show()
```

## Evolution and Convergence Analysis

Now let's run the genetic algorithm and track how the population evolves.

```python
# Track fitness over generations
best_fitness_history = []
mean_fitness_history = []
std_fitness_history = []

# Run evolution manually to capture statistics
print("Running evolution and collecting statistics...\n")

for gen in range(50):
    # Get current population fitness
    current_fitness = [chrom.fitness() for chrom in manager.get_chrom_array()]
    
    best_fitness_history.append(max(current_fitness))
    mean_fitness_history.append(np.mean(current_fitness))
    std_fitness_history.append(np.std(current_fitness))
    
    # Evolve one generation
    if gen < 49:
        manager.greedy_next_generation()
        manager._EvolManager__current_generation += 1
    
    if gen % 10 == 0:
        print(f"Generation {gen:3d}: Best={best_fitness_history[-1]:.4f}, "
              f"Mean={mean_fitness_history[-1]:.4f}, "
              f"Std={std_fitness_history[-1]:.4f}")

print("\nEvolution complete!")
```

```python
# Plot convergence
fig, ax = plt.subplots(figsize=(12, 6))

generations = range(len(best_fitness_history))

# Plot best and mean fitness
ax.plot(generations, best_fitness_history, linewidth=2, color='#2E86AB', label='Best Fitness')
ax.plot(generations, mean_fitness_history, linewidth=2, color='#A23B72', label='Mean Fitness')

# Add standard deviation band
mean_arr = np.array(mean_fitness_history)
std_arr = np.array(std_fitness_history)
ax.fill_between(generations, mean_arr - std_arr, mean_arr + std_arr, 
                alpha=0.2, color='#A23B72', label='±1 Std Dev')

ax.set_xlabel('Generation', fontsize=12, weight='bold')
ax.set_ylabel('Fitness', fontsize=12, weight='bold')
ax.set_title('Genetic Algorithm Convergence (12-Queens, Pop=1000)', fontsize=14, weight='bold')
ax.legend(loc='lower right', fontsize=11)
ax.grid(True, alpha=0.3, linestyle='--')
ax.set_ylim(0, 1.05)

plt.tight_layout()
plt.show()
```

## Final Population Analysis

Compare the final population to the initial population.

```python
# Get final population
final_population = manager.get_chrom_array()
final_fitness = [chrom.fitness() for chrom in final_population]

print(f"Final Population Statistics (After 50 generations)")
print(f"{'='*60}")
print(f"Mean fitness:   {np.mean(final_fitness):.4f}")
print(f"Std deviation:  {np.std(final_fitness):.4f}")
print(f"Min fitness:    {np.min(final_fitness):.4f}")
print(f"Max fitness:    {np.max(final_fitness):.4f}")
print(f"Median fitness: {np.median(final_fitness):.4f}")
print(f"\nImprovement:")
print(f"Mean fitness change: {np.mean(final_fitness) - np.mean(initial_fitness):.4f}")
print(f"Best fitness change: {np.max(final_fitness) - np.max(initial_fitness):.4f}")
```

```python
# Compare initial vs final distributions
fig, (ax1, ax2) = plt.subplots(1, 2, figsize=(14, 5))

# Initial distribution
sns.kdeplot(data=initial_fitness, ax=ax1, fill=True, color='#A23B72', alpha=0.5, label='Initial')
ax1.axvline(np.mean(initial_fitness), color='#A23B72', linestyle='--', linewidth=2)
ax1.set_xlabel('Fitness', fontsize=12, weight='bold')
ax1.set_ylabel('Density', fontsize=12, weight='bold')
ax1.set_title('Initial Population (Generation 0)', fontsize=14, weight='bold')
ax1.set_xlim(0, 1)
ax1.grid(True, alpha=0.3)

# Final distribution
sns.kdeplot(data=final_fitness, ax=ax2, fill=True, color='#2E86AB', alpha=0.5, label='Final')
ax2.axvline(np.mean(final_fitness), color='#2E86AB', linestyle='--', linewidth=2)
ax2.set_xlabel('Fitness', fontsize=12, weight='bold')
ax2.set_ylabel('Density', fontsize=12, weight='bold')
ax2.set_title('Final Population (Generation 50)', fontsize=14, weight='bold')
ax2.set_xlim(0, 1)
ax2.grid(True, alpha=0.3)

plt.tight_layout()
plt.show()
```

## Solutions Analysis

Analyze the solutions found during evolution.

```python
# Get solution data
solutions = manager.get_solutions()
solution_generations = manager.get_sol_generations()
solution_times = manager.get_solution_times()

print(f"Solutions Found: {len(solutions)}")
print(f"{'='*60}")

if len(solutions) > 0:
    print(f"First solution found: Generation {solution_generations[0]}")
    print(f"Last solution found:  Generation {solution_generations[-1]}")
    print(f"\nSolution discovery rate:")
    
    # Group by generation ranges
    early = sum(1 for g in solution_generations if g < 17)
    mid = sum(1 for g in solution_generations if 17 <= g < 34)
    late = sum(1 for g in solution_generations if g >= 34)
    
    print(f"  Generations 0-16:   {early} solutions ({early/len(solutions)*100:.1f}%)")
    print(f"  Generations 17-33:  {mid} solutions ({mid/len(solutions)*100:.1f}%)")
    print(f"  Generations 34-50:  {late} solutions ({late/len(solutions)*100:.1f}%)")
```

```python
# Visualize solution discovery
if len(solutions) > 0:
    fig, (ax1, ax2) = plt.subplots(1, 2, figsize=(14, 5))
    
    # Solutions over generations
    ax1.scatter(solution_generations, range(1, len(solution_generations)+1), 
               s=100, color='#2E86AB', alpha=0.6, edgecolors='black', linewidth=1.5)
    ax1.set_xlabel('Generation', fontsize=12, weight='bold')
    ax1.set_ylabel('Cumulative Solutions', fontsize=12, weight='bold')
    ax1.set_title('Solution Discovery by Generation', fontsize=14, weight='bold')
    ax1.grid(True, alpha=0.3, linestyle='--')
    
    # Solutions over time
    ax2.scatter(solution_times, range(1, len(solution_times)+1),
               s=100, color='#A23B72', alpha=0.6, edgecolors='black', linewidth=1.5)
    ax2.set_xlabel('Time (seconds)', fontsize=12, weight='bold')
    ax2.set_ylabel('Cumulative Solutions', fontsize=12, weight='bold')
    ax2.set_title('Solution Discovery by Time', fontsize=14, weight='bold')
    ax2.grid(True, alpha=0.3, linestyle='--')
    
    plt.tight_layout()
    plt.show()
```

## Performance Across Board Sizes

Compare algorithm performance for different N values.

```python
# Test different board sizes
board_sizes = [8, 10, 12]
results = []

print("Testing different board sizes...\n")

for n in board_sizes:
    test_manager = EvolManager(
        genes_per_chrom=n,
        pop=500,
        generations=30
    )
    
    # Track best fitness
    best_fitness = []
    for gen in range(30):
        current_pop = test_manager.get_chrom_array()
        best_fitness.append(max(chrom.fitness() for chrom in current_pop))
        if gen < 29:
            test_manager.greedy_next_generation()
            test_manager._EvolManager__current_generation += 1
    
    solutions_found = len(test_manager.get_solutions())
    
    results.append({
        'N': n,
        'Solutions': solutions_found,
        'Final_Best': best_fitness[-1],
        'Convergence': best_fitness
    })
    
    print(f"{n}-Queens: {solutions_found} solutions, Final best fitness: {best_fitness[-1]:.4f}")

print("\nDone!")
```

```python
# Plot convergence comparison
fig, ax = plt.subplots(figsize=(12, 6))

colors = ['#2E86AB', '#A23B72', '#F18F01']
for i, result in enumerate(results):
    ax.plot(range(30), result['Convergence'], 
           linewidth=2, color=colors[i], label=f"{result['N']}-Queens")

ax.set_xlabel('Generation', fontsize=12, weight='bold')
ax.set_ylabel('Best Fitness', fontsize=12, weight='bold')
ax.set_title('Convergence Comparison Across Board Sizes', fontsize=14, weight='bold')
ax.legend(loc='lower right', fontsize=11)
ax.grid(True, alpha=0.3, linestyle='--')
ax.set_ylim(0, 1.05)

plt.tight_layout()
plt.show()
```
